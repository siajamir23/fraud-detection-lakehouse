```python

# ==========================================
# 00 - ENVIRONMENT SETUP
# Fraud Detection Lakehouse
# ==========================================

# 1. Project configuration

PROJECT_NAME = "fraud_detection"
CATALOG_NAME = "fraud_detection"

SCHEMAS = [
    "bronze",
    "silver",
    "gold",
    "quarantine"
]

# 2. Create catalog if it does not exist

spark.sql(f"CREATE CATALOG IF NOT EXISTS {CATALOG_NAME}")

# 3. Create project schemas

for schema in SCHEMAS:
    spark.sql(f"""
        CREATE SCHEMA IF NOT EXISTS {CATALOG_NAME}.{schema}
    """)

# 4. Set active catalog

spark.sql(f"USE CATALOG {CATALOG_NAME}")

# 5. Verify configuration

print(f"Project: {PROJECT_NAME}")
print(f"Current catalog: {spark.sql('SELECT current_catalog()').first()[0]}")

display(
    spark.sql(f"SHOW SCHEMAS IN {CATALOG_NAME}")
)
```
